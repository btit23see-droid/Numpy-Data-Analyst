### What is numpy?

NumPy is a Python library mainly used for scientific and numerical computing. Its core feature is the ndarray object, which allows us to store data in one-dimensional or multi-dimensional arrays. These arrays can handle large amounts of data efficiently and all elements usually have the same data type. NumPy also provides fast functions for mathematical and other array operations.

### Numpy Arrays Vs Python Sequences

NumPy arrays have a fixed size when they are created. Unlike Python lists, they cannot be resized directly; changing their size creates a new array.
. All elements in a NumPy array have the same data type, so they generally use the same amount of memory.
. NumPy makes mathematical operations easier and faster when working with large amounts of data, often requiring less code than Python lists.
. Many scientific and mathematical Python libraries use NumPy arrays. They often convert Python lists into NumPy arrays for processing and may also return NumPy arrays as output.

### np.ceil() → Ceiling

Definition: Rounds each number UP to the nearest integer.

Always goes toward +∞, no matter what.

```python
import numpy as np

arr = np.array([1.2, 2.7, -1.5, -2.3])
arr
```

```python
np.ceil(arr)
```

### np.floor() → Floor

Definition: Rounds each number DOWN to the nearest integer.

Always goes toward -∞.

```python
np.floor(arr)
```

### np.round() → Round (to nearest)

Definition: Rounds each number to the nearest integer.

If decimal = 0.5 → rounds to nearest even integer (NumPy behavior).

```python
np.round(arr)
```
